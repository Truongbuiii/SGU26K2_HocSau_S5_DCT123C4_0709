# **Bài 4. Slicing và Indexing**

Truy xuất một phần dữ liệu từ Tensor nhiều chiều.

```python
import torch

# 1. Code mẫu từ đề bài
a = torch.randn(3, 4, 5)
print("Original tensor shape:", a.shape)
print("Original tensor:\n", a)

print("\nTensor a[1]:\n", a[1])
print("Shape a[1]:", a[1].shape)

print("\nTensor a[1:, 2:4]:\n", a[1:, 2:4])
print("Shape a[1:, 2:4]:", a[1:, 2:4].shape)

# 2. Thực hiện yêu cầu: Lấy hai phần tử cuối cùng trên chiều thứ ba
# Dùng slicing [..., -2:] hoặc [:, :, -2:]
last_two_elements = a[:, :, -2:]
print("\nHai phan tu cuoi tren chieu thu ba (shape):", last_two_elements.shape)
print("Gia tri:\n", last_two_elements)
```

Yêu cầu:
• Cho biết shape của a[1] và a[1:, 2:4].
• Thử lấy hai phần tử cuối cùng trên chiều thứ ba.

Hình dạng (shape) của các lát cắt (slicing):   
- a[1]: Có shape là torch.Size([4, 5]). Khi cố định chỉ số ở chiều thứ nhất (index = 1), chiều này bị tiêu biến, để lại ma trận 2 chiều gồm 4 hàng và 5 cột.
- a[1:, 2:4]: Có shape là torch.Size([2, 2, 5]). Ta cắt lấy các khối từ chỉ số 1 trở đi (chiều 0 còn 2 khối), lấy các hàng từ 2 đến trước 4 (chiều 1 còn 2 hàng), và giữ nguyên toàn bộ 5 cột ở chiều cuối cùng.

Lấy hai phần tử cuối cùng trên chiều thứ ba:   Cú pháp: a[:, :, -2:] (hoặc a[..., -2:]).
- Ý nghĩa: Giữ nguyên tất cả các khối (chiều 0) và tất cả các hàng (chiều 1), chỉ lấy 2 phần tử cuối cùng ở chiều cột (chiều 2) thông qua chỉ số âm -2:. Kết quả trả về một Tensor có kích thước torch.Size([3, 4, 2]).

# **Bài 6. Phép toán và nhân ma trận**
Thực hiện các phép toán phần tử và phép nhân ma trận bằng toán tử @.

```python
import torch

a = torch.randn(2, 3)
b = torch.randn(2, 3)

print("a:\n", a)
print("b:\n", b)
print("\na + b:\n", a + b)
print("\na^2:\n", a ** 2)
print("\na / b:\n", a / b)

# Nhan tung phan tu vs Nhan ma tran
elementwise_prod = a * b
matrix_mul = a @ b.T

print("\nElement-wise multiplication (a * b):\n", elementwise_prod)
print("\nMatrix multiplication (a @ b.T):\n", matrix_mul)
print("Shape cua a @ b.T:", matrix_mul.shape)
```

# Trả lời yêu cầu
Phân biệt a * b và a @ b.T
- a * b : phép nhân từng phần tử tương ứng giữa 2 tensor cùng kích thước (2x3), kết quả ra shape (2,3)
- a @ b.T : Phép nhân ma trận đại số tuyến tính. Ma trận a kích thước (2 x 3) nhân với chuyển vị của b kích thước (3x2)

Shape của kết quả a @ b.T là torch.Size([2, 2])

# **Bài 7. Các hàm thống kê trên Tensor**

Tính trung bình, độ lệch chuẩn và tổng tích lũy theo từng cột.

# **Bài 8. Autograd cơ bản**

Tạo biến có requires_grad=True và dùng backward() để tính đạo hàm tự động.

```python
import torch

# 1. Ham y = x * x
x = torch.tensor(3.6, requires_grad=True)
y = x * x
y.backward()
print("x =", x)
print("y =", y)
print("x.grad =", x.grad)

# 2. Yeu cau: Doi thanh y = 3*x**2 + 2*x + 1
x2 = torch.tensor(3.6, requires_grad=True)
y2 = 3 * x2**2 + 2 * x2 + 1
y2.backward()
print("\n--- Ham y = 3*x^2 + 2*x + 1 ---")
print("x2.grad =", x2.grad)
```

# Trả lời yêu cầu:

1. Tính thủ công đạo hàm của y = x2 tại x = 3.6 và so sánh với x.grad.

   Đạo hàm thủ công hàm y = x2 tại x = 3.6

       y'(x) = 2x => y'(3.6) = 2 x 3.6 = 7.2

  -> Giá trị trùng khớp hoàn toàn với x.grad thu được từ pytorch(7.2000)

2. Đổi hàm thành y = 3*x**2 + 2*x + 1 và kiểm tra gradient.

   Đạo hàm thủ công hàm y = 3*x**2 + 2*x + 1 tại x = 3.6

       y'(x) = 6x + 2 => y'(3.6) = 6 x 3.6 + 2 = 23.6
   
   Kết quả autograd x2.grad tính ra chính xác 23.6000

# **Bài 9. Gradient Descent tìm hệ số đa thức**

Sinh dữ liệu từ đa thức y = x2 + 2x + 3, sau đó dùng NAdam và autograd để tìm lại ba hệ số.

```python
import numpy as np
import torch

polynomial = np.poly1d([1, 2, 3])
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)

XX = np.hstack([X * X, X, np.ones_like(X)])
w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)

optimizer = torch.optim.NAdam([w], lr=0.01)

print("Initial coefficients:\n", w.data)

for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()

print("\nLearned coefficients:\n", w.data)
```

Yêu cầu:

1. So sánh ba hệ số học được với [1, 2, 3] : Sau 1000 epoch huấn luyện, các hệ số trong ma trận $w$ hội tụ rất sát với các giá trị thực tế $A \approx 1$, $B \approx 2$, $C \approx 3$.   

2. Giải thích vai trò của optimizer.zero_grad(), mse.backward() và optimizer.step()?
- optomizer.zero_grad(): xóa sạch các gradient tích lũy từ các vòng lặp trước đó về 0 trước khi tính đạo hàm mới
- mse.backward(): Kích hoạt lan truyền ngược để tính toán gradient của hàm mất mát mse đối với trọng số $w$($\frac{\partial \text{mse}}{\partial w}$).
- optimizer.step(): Cập nhật giá trị các trọng số $w$ theo hướng giảm dần của gradient dựa vào thuật toán tối ưu NAdam

# **Bài 10. Giải hệ phương trình bằng Autograd**

Tối thiểu hóa tổng sai số bình phương để tìm A, B, C, D thỏa bốn phương trình.

```python
import random
import torch

A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)

EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)

for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2
    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4
    
    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()

print(f"A = {A.item():.4f}")
print(f"B = {B.item():.4f}")
print(f"C = {C.item():.4f}")
print(f"D = {D.item():.4f}")

# Kiem tra tong sai so cuoi cung
print("Final loss (sqerr):", sqerr.item())
```

Trả lời yêu cầu:

• Kiểm tra nghiệm thu được với bốn phương trình: A+B=9, C-D=1, A+C=8, B-D=2.
- $A + B \approx 9$

- $C - D \approx 1$

- $A + C \approx 8$

- $B - D \approx 2$


-> Hàm mất mát sqerr tiệm cận về 0, chứng minh nghiệm tìm được hoàn toàn thỏa mãn.

• Nhận xét vì sao bài toán có thể có nhiều nghiệm:

Coi hệ 4 phương trình tuyến tính 4 ẩn:

$$\begin{cases} A + B = 9 \\ C - D = 1 \\ A + C = 8 \\ B - D = 2 \end{cases}$$

Lấy phương trình (1) trừ phương trình (4):
$(A + B) - (B - D) = 9 - 2 \implies A + D = 7$.

Cộng phương trình (2) và (3): $(C - D) + (A + C) = 1 + 8 \implies A + 2C - D = 9$.

Thực chất phương trình (4) phụ thuộc tuyến tính vào 3 phương trình còn lại (hệ có hạng ma trận hệ số bằng 3 nhưng có 4 ẩn). Do đó hệ có bậc tự do bằng 1, dẫn đến việc có vô số nghiệm phụ thuộc vào giá trị khởi tạo ngẫu nhiên ban đầu.

# Bài 11. Viết chương trình PyTorch để thực hiện các yêu cầu sau:

```python
import numpy as np
import torch

# Yeu cau 1: Tao Tensor X co 24 phan tu va reshape
print("=== YEU CAU 1 ===")
X_tensor = torch.arange(24)
print("Reshape 4x6:\n", X_tensor.reshape(4, 6))
print("Reshape 2x12:\n", X_tensor.reshape(2, 12))
print("Reshape 2x3x4:\n", X_tensor.reshape(2, 3, 4))

# Yeu cau 2: Thao tac tren Tensor A, B kich thuoc 3x4
print("\n=== YEU CAU 2 ===")
A = torch.randn(3, 4)
B = torch.randn(3, 4)
print("A + B:\n", A + B)
print("A * B:\n", A * B)
print("A / B:\n", A / B)
print("Trung binh A theo cot (dim=0):", torch.mean(A, dim=0))
print("Do lech chuan A theo cot (dim=0):", torch.std(A, dim=0))

# Yeu cau 3: Tinh dy/dx cua y = 2x^2 + 5x + 3 tai x = 2
print("\n=== YEU CAU 3 ===")
x_val = torch.tensor(2.0, requires_grad=True)
y_func = 2 * (x_val ** 2) + 5 * x_val + 3
y_func.backward()
print("Autograd dy/dx tai x = 2:", x_val.grad.item())
# Tinh thu cong: dy/dx = 4x + 5 => 4*2 + 5 = 13
print("Tinh thu cong dy/dx: 4*(2) + 5 = 13")

# Yeu cau 4: Huan luyen da thuc y = 2x^2 - 3x + 5
print("\n=== YEU CAU 4 ===")
poly_new = np.poly1d([2, -3, 5])
N = 30
X_poly = np.random.randn(N, 1) * 3
Y_poly = poly_new(X_poly)

XX_poly = np.hstack([X_poly * X_poly, X_poly, np.ones_like(X_poly)])
w_poly = torch.randn(3, 1, requires_grad=True)
x_p = torch.tensor(XX_poly, dtype=torch.float32)
y_p = torch.tensor(Y_poly, dtype=torch.float32)

opt = torch.optim.NAdam([w_poly], lr=0.01)

for epoch in range(1500):
    opt.zero_grad()
    pred = x_p @ w_poly
    loss = torch.mean(torch.square(y_p - pred))
    loss.backward()
    opt.step()

print("He so tim duoc sau khi huan luyen:")
print(f"A = {w_poly[0].item():.4f} (Goc: 2)")
print(f"B = {w_poly[1].item():.4f} (Goc: -3)")
print(f"C = {w_poly[2].item():.4f} (Goc: 5)")
```

Trả lời yêu cầu:

- Đạo hàm thủ công yêu cầu 3:
$$y'(x) = 4x + 5 \implies y'(2) = 4(2) + 5 = 13$$

Trùng khớp với kết quả autograd trả về $13.0$.   

- Kết quả huấn luyện yêu cầu 4:
Mô hình hội tụ về các hệ số xấp xỉ $[2, -3, 5]$, đúng với đa thức khởi tạo $y = 2x^2 - 3x + 5$.
